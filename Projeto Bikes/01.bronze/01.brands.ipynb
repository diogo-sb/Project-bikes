```python
#Lendo o arquivo csv
df_brands = spark.read.format('csv')\
.option('header','true')\
.option('inferSchema','true')\
.load(resource_path + 'brands.csv')
```

```python

display(df_brands)
```

```python
df_brands.write.mode('overwrite')\
.format('delta')\
.option('mergeSchema','true')\
.save(bronze_path + 'brand')
```
